```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('cleaned_funding_rounds.csv')
df['Date'] = pd.to_datetime(df['Date'])

print(df.shape)
df.head()
```

```python
pune = df[df['City'] == 'Pune']['InvestmentAmount_USD']
delhi = df[df['City'] == 'Delhi']['InvestmentAmount_USD']

print("Pune median:", pune.median(), "| Pune mean:", pune.mean(), "| n =", len(pune))
print("Delhi median:", delhi.median(), "| Delhi mean:", delhi.mean(), "| n =", len(delhi))

stat, p_value = stats.mannwhitneyu(pune, delhi, alternative='two-sided')
print("\nMann-Whitney U statistic:", stat)
print("p-value:", p_value)
```

```python
early = df[df['Stage_Group'] == 'Early']['InvestmentAmount_USD']
late = df[df['Stage_Group'] == 'Late']['InvestmentAmount_USD']

print("Early median:", early.median(), "| n =", len(early))
print("Late median:", late.median(), "| n =", len(late))

stat2, p_value2 = stats.mannwhitneyu(early, late, alternative='two-sided')
print("\np-value:", p_value2)
```

```python
correlation, p_val_corr = stats.spearmanr(df['Year'], df['Log_Amount'])
print("Spearman correlation:", correlation)
print("p-value:", p_val_corr)
```

```python
foodtech = df[df['Industry'] == 'FoodTech']['InvestmentAmount_USD']

mean_ft = foodtech.mean()
sem_ft = stats.sem(foodtech)  # standard error of the mean
ci = stats.t.interval(0.95, len(foodtech)-1, loc=mean_ft, scale=sem_ft)

print("FoodTech average deal size:", round(mean_ft))
print("95% Confidence Interval:", (round(ci[0]), round(ci[1])))
```
